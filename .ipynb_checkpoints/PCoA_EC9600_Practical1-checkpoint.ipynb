# Principal Coordinates Analysis — from scratch

**EC9600 · Practical Assignment 1 · Group GX**

---

### What this notebook does

We implement Principal Coordinates Analysis (PCoA, also called *classical* or *metric*
multidimensional scaling) from first principles, then apply it to a 675-sample microbiome
dataset and evaluate the result statistically, visually and computationally.

PCoA answers a question PCA cannot: *given only the pairwise dissimilarities between
objects — not their coordinates — where should we place them in a low-dimensional space so
that the distances between the placed points reproduce those dissimilarities as closely as
possible?* That matters in microbial ecology because the dissimilarities people actually
care about (Bray–Curtis, UniFrac, Jaccard) are not Euclidean distances in the OTU space, so
PCA on the abundance table answers the wrong question.

### The algorithm

Given an $N \times N$ dissimilarity matrix $\mathbf{D}$ (Gower 1966; restated in
Shi *et al.* 2020, the source of this dataset):

| Step | Operation |
|---|---|
| 1 | $\mathbf{A} = [a_{hi}]$ where $a_{hi} = -\tfrac{1}{2} D_{hi}^{2}$ |
| 2 | $\mathbf{G} = \left(\mathbf{I} - \tfrac{\mathbf{1}\mathbf{1}'}{N}\right)\mathbf{A}\left(\mathbf{I} - \tfrac{\mathbf{1}\mathbf{1}'}{N}\right)$ — Gower's doubly-centred matrix |
| 3 | $\mathbf{G} = \mathbf{U}\boldsymbol{\Lambda}\mathbf{U}'$ — symmetric eigendecomposition |
| 4 | $\mathbf{Y}_k = \mathbf{U}_k\boldsymbol{\Lambda}_k^{1/2}$ — project onto the $k$ leading axes |

The logic behind step 1–2: if $\mathbf{D}$ *were* the Euclidean distance matrix of some
centred configuration $\mathbf{Y}_c$, then $-\tfrac12 D_{ij}^2$ doubly centred is exactly
the Gram matrix $\mathbf{Y}_c\mathbf{Y}_c'$. So PCoA inverts the map
"coordinates → distances": it recovers a Gram matrix from distances and factorises it.
When $\mathbf{D}$ is *not* Euclidean-embeddable, $\mathbf{G}$ is no longer positive
semi-definite and some eigenvalues come out **negative** — those dimensions would need
imaginary coordinates. Section 6 measures and fixes this.

### 1.2 Audit: what is actually in this matrix?

Never trust the shape alone. The audit below drives every preprocessing decision that
follows.

````python
audit = validate_table(X_raw)
show_table(pd.DataFrame({"property": list(audit), "value": [audit[k] for k in audit]}),
           "data_audit")
````

The audit exposes the single most important property of this table:

* rows sum to 1 — these are **closed relative abundances**, i.e. compositional data;
* there is **not one exact zero**, yet 97.6% of entries sit below $10^{-9}$.

Those sub-$10^{-9}$ entries are drawn from a uniform distribution on $(0, 10^{-10}]$: the
authors applied **zero replacement** so that log-ratio methods would be defined. They are
structural zeros wearing a disguise. If we leave them alone, a presence/absence metric like
Jaccard concludes that every taxon is present in every sample and returns a distance matrix
of zeros. The histogram below shows the two populations cleanly separated by ~6 orders of
magnitude, which is what makes a threshold defensible rather than arbitrary.

````python
fig, axes = plt.subplots(1, 3, figsize=(13.5, 3.6))

logv = np.log10(X_raw[X_raw > 0].ravel())
axes[0].hist(logv, bins=120, color="#4C72B0")
axes[0].axvline(-9, color="#C44E52", ls="--", lw=1.6, label="threshold $10^{-9}$")
axes[0].set(xlabel="$\\log_{10}$(relative abundance)", ylabel="entries",
            title="Bimodal: real signal vs zero-replacement noise")
axes[0].legend()

prev = (X_raw > 1e-9).sum(axis=0)
axes[1].hist(prev, bins=80, color="#55A868")
axes[1].axvline(5, color="#C44E52", ls="--", lw=1.6, label="prevalence filter (5)")
axes[1].set(xlabel="samples in which a taxon is observed", ylabel="taxa",
            title="Most taxa are near-absent", yscale="log")
axes[1].legend()

order = np.sort(X_raw.max(axis=0))[::-1]
axes[2].plot(np.arange(1, len(order) + 1), np.cumsum(order) / order.sum(),
             color="#8172B3", lw=1.8)
axes[2].axhline(0.99, color="grey", ls=":", lw=1.2)
axes[2].set(xlabel="taxa ranked by peak abundance", ylabel="cumulative share of peak mass",
            title="A minority of taxa carry the signal", xscale="log")

plt.tight_layout(); save(fig, "fig01_data_audit"); plt.show()

RESULTS["fraction_at_zero_floor"] = audit["fraction_at_zero_floor"]
RESULTS["n_taxa_never_observed"] = audit["n_taxa_never_observed"]
````

### 1.3 Preprocessing

Different metrics want different things from the same table, so we build **two views**:

| View | Zeros | Used by | Why |
|---|---|---|---|
| `X_count` | restored to exact 0 | Bray–Curtis, Jaccard, Hellinger, Jensen–Shannon | these measure abundance *differences*; the $10^{-11}$ noise is meaningless to them and makes Jaccard degenerate |
| `X_positive` | kept as replaced | Aitchison (clr) | $\log 0$ is undefined; the replacement values are precisely what makes the log-ratio transform possible |

Both views drop taxa observed in fewer than 5 samples. That removes 64% of the columns but
only 0.7% of the abundance mass — rare taxa contribute almost nothing to Bray–Curtis while
adding variance, and they dominate log-ratio metrics, where a taxon's *replacement value* can
swing the clr vector. After filtering we re-close each row to sum to 1 so Bray–Curtis stays
comparable across samples.

````python
MIN_PREVALENCE = 5

X_count, taxa, report_count = preprocess(
    X_raw, data.taxa, min_prevalence=MIN_PREVALENCE, unzero=True, renormalise=True)
X_positive, _, _ = preprocess(
    X_raw, data.taxa, min_prevalence=MIN_PREVALENCE, unzero=False, renormalise=True)

print(f"taxa retained      : {report_count['n_taxa_kept']} / {report_count['n_taxa_in']} "
      f"({report_count['fraction_taxa_kept']:.1%})")
print(f"abundance retained : {report_count['mean_retained_mass']:.4f} mean, "
      f"{report_count['min_retained_mass']:.4f} worst-case sample")
print(f"true sparsity      : {(X_count == 0).mean():.1%} of the filtered matrix is zero")

RESULTS["preprocessing"] = report_count
RESULTS["min_prevalence"] = MIN_PREVALENCE
````

## 2 · The algorithm, on a problem whose answer we already know

Before running anything on 675 samples we verify the four steps on a unit square. Its
distance matrix has an exactly known solution: two equal non-zero eigenvalues (the square is
symmetric under 90° rotation), everything else zero, and the recovered coordinates must be
the square itself up to rotation and reflection — PCoA cannot recover orientation, only
shape, because distances are invariant to rigid motion.

````python
square = np.array([[0., 0.], [1., 0.], [1., 1.], [0., 1.]])
D_sq = pairwise_distances(square, metric="euclidean")

A_sq = -0.5 * D_sq ** 2                      # step 1
G_sq = gower_center(A_sq)                    # step 2
eigvals, eigvecs = np.linalg.eigh(G_sq)      # step 3
order = np.argsort(eigvals)[::-1]
eigvals, eigvecs = eigvals[order], eigvecs[:, order]
Y_sq = eigvecs[:, :2] * np.sqrt(eigvals[:2]) # step 4

print("step 1  A = -0.5 * D^2\n", np.round(A_sq, 3))
print("\nstep 2  G = C A C   (rows and columns now sum to zero: "
      f"max |rowsum| = {np.abs(G_sq.sum(1)).max():.1e})\n", np.round(G_sq, 3))
print("\nstep 3  eigenvalues:", np.round(eigvals, 6))
print("\nstep 4  coordinates\n", np.round(Y_sq, 3))

result_sq = pcoa.pcoa(D_sq, n_components=2)
print(f"\nreconstruction error vs original D : "
      f"{np.abs(result_sq.reconstructed_distances() - D_sq).max():.2e}")
print(f"Procrustes m^2 against the true square : {procrustes_m2(Y_sq, square):.2e}")
````

````python
fig, axes = plt.subplots(1, 2, figsize=(8.6, 3.8))
for ax, P, ttl in [(axes[0], square, "Input configuration"),
                   (axes[1], Y_sq, "Recovered by PCoA (from distances only)")]:
    ax.scatter(P[:, 0], P[:, 1], s=110, c=["#4C72B0", "#DD8452", "#55A868", "#C44E52"],
               zorder=3, edgecolor="white", lw=1)
    for i, (x, y) in enumerate(P):
        ax.annotate(f"P{i+1}", (x, y), xytext=(6, 6), textcoords="offset points", fontsize=9)
    loop = np.vstack([P, P[:1]])
    ax.plot(loop[:, 0], loop[:, 1], color="grey", lw=1, alpha=0.6, zorder=1)
    ax.set_title(ttl); ax.set_aspect("equal", adjustable="datalim")
plt.tight_layout(); save(fig, "fig02_toy_example"); plt.show()
````

The recovered square is rotated 45° and reflected relative to the input — as it must be
— but the shape and every pairwise distance are exact to machine precision. This is the
minimal proof that steps 1–4 are implemented correctly; the full test suite in section 10
extends it to 36 cases.

**Complexity.** Step 3 dominates: a dense symmetric eigendecomposition is $O(N^3)$ in time
and $O(N^2)$ in memory, independent of the number of taxa. The *distance* computation is
$O(N^2 P)$, which for this dataset ($N{=}675$, $P{=}2385$) is the more expensive half — see
section 8.

## 3 · PCoA of the microbiome data

Headline analysis: **Bray–Curtis** dissimilarity on the filtered, re-closed table. It is the
standard in microbial ecology and the measure used in the source paper. One useful identity
for this dataset: because every row sums to 1, the Bray–Curtis denominator
$\sum_k (x_{ik} + x_{jk})$ is always 2, so here Bray–Curtis is exactly half the $L_1$
distance. We assert that below rather than assuming it.

````python
t0 = time.perf_counter()
D_bc = pairwise_distances(X_count, metric="braycurtis", memory_budget_mb=256)
time_bc = time.perf_counter() - t0

from scipy.spatial.distance import pdist, squareform
assert np.allclose(D_bc, 0.5 * squareform(pdist(X_count, "cityblock")), atol=1e-12)

print(f"Bray-Curtis matrix computed in {time_bc:.1f} s   ({D_bc.nbytes/1e6:.1f} MB)")
print(f"dissimilarity range: {D_bc[np.triu_indices_from(D_bc,1)].min():.3f} "
      f"- {D_bc[np.triu_indices_from(D_bc,1)].max():.3f}, "
      f"mean {D_bc[np.triu_indices_from(D_bc,1)].mean():.3f}")
RESULTS["braycurtis_time_s"] = time_bc
````

````python
t0 = time.perf_counter()
res_bc = pcoa.pcoa(D_bc, n_components=10)
time_pcoa = time.perf_counter() - t0
print(f"PCoA in {time_pcoa:.3f} s\n")
print(res_bc.summary(n_axes=6))

RESULTS["pcoa_time_s"] = time_pcoa
RESULTS["negative_fraction"] = res_bc.negative_fraction
RESULTS["n_negative"] = res_bc.n_negative
RESULTS["explained_top2"] = float(res_bc.proportion_explained[:2].sum())
RESULTS["explained_top3"] = float(res_bc.proportion_explained[:3].sum())
````

Two numbers deserve comment.

**14.4% + 13.5% on the first two axes.** That looks low next to a typical PCA, but it is
entirely normal for microbiome ordination: with thousands of taxa the total variation is
spread across hundreds of genuine dimensions. The right question is not "is 28% high?" but
"do the leading axes carry *structure* rather than noise?" — the broken-stick reference on
the scree plot and the PERMANOVA in section 7 answer that.

**364 negative eigenvalues, 11.0% of the absolute trace.** Bray–Curtis is not
Euclidean-embeddable, so $\mathbf{G}$ is indefinite. Roughly a ninth of the dissimilarity
structure cannot be represented in *any* real-coordinate space. That is a property of the
metric, not a bug; section 6 quantifies what it costs and what the fixes cost in return.

````python
fig, ax = plt.subplots(figsize=(7.2, 3.8))
scree_plot(res_bc, n_axes=20, ax=ax)
plt.tight_layout(); save(fig, "fig03_scree"); plt.show()

res_full = pcoa.pcoa(D_bc)          # every positive axis, for the fidelity analysis below
pos = res_full.eigenvalues[res_full.eigenvalues > 0]
m = pos.size
bs = np.array([np.sum(1.0 / np.arange(i, m + 1)) / m for i in range(1, m + 1)])
below = np.flatnonzero((pos / pos.sum()) < bs)
n_keep = int(below[0]) if below.size else m
print(f"{m} positive axes in total")
print(f"broken-stick rule retains the first {n_keep} axes "
      f"({100*res_full.proportion_explained[:n_keep].sum():.1f}% of the positive variation)")
RESULTS["n_positive_axes"] = int(m)
RESULTS["broken_stick_axes"] = n_keep
RESULTS["broken_stick_explained"] = float(res_full.proportion_explained[:n_keep].sum())
````

The heat map is the un-projected view of exactly the same information the ordination
summarises: block structure along the diagonal means within-donor samples are more similar to
each other than to samples from other donors. Seeing the blocks in the raw dissimilarity
matrix confirms the pattern is in the data, not an artefact of the projection.

The biplot arrows are *weighted-average* projections. PCoA works from a distance matrix, so
taxa have no loadings of their own; we correlate each taxon's abundance profile with the
sample scores and draw the strongest. An arrow points towards the region of the plot where
that taxon is enriched. These arrows are a descriptive aid, not a statistical test — the
named taxa are candidates for follow-up differential-abundance testing, nothing more.

## 5 · Does the choice of dissimilarity matter?

PCoA is only as meaningful as the dissimilarity fed to it, so the metric is a modelling
decision, not a default. Six measures, same pipeline:

| Metric | What it responds to | Euclidean-embeddable? |
|---|---|---|
| Bray–Curtis | abundance differences, scaled by total abundance | no |
| Jaccard | presence/absence only — ignores abundance | no (but nearly, in practice) |
| Hellinger | $\sqrt{}$-transformed abundance; down-weights dominant taxa | yes |
| Aitchison | log-ratios; the scale-invariant choice for compositions | yes |
| Jensen–Shannon | divergence between abundance profiles as distributions | yes |
| Euclidean | raw abundance differences (PCoA here $\equiv$ PCA) | yes |

````python
metric_specs = [
    ("braycurtis",    X_count,    {}),
    ("jaccard",       X_count,    {"presence_threshold": 0.0}),
    ("hellinger",     X_count,    {}),
    ("aitchison",     X_positive, {}),
    ("jensenshannon", X_count,    {}),
    ("euclidean",     X_count,    {}),
]

distances, results_by_metric, rows = {}, {}, []
for name, Xin, kw in metric_specs:
    t0 = time.perf_counter()
    D = pairwise_distances(Xin, metric=name, **kw)
    elapsed = time.perf_counter() - t0
    r = pcoa.pcoa(D, n_components=10)
    distances[name], results_by_metric[name] = D, r
    rows.append({
        "metric": name,
        "compute (s)": elapsed,
        "neg. eigenvalue %": 100 * r.negative_fraction,
        "PCo1 %": 100 * r.proportion_explained[0],
        "PCo1+2 %": 100 * r.proportion_explained[:2].sum(),
        "stress (2 axes)": kruskal_stress(D, r.reconstructed_distances(2)),
        "Donor R^2": permanova(D, metadata["Donor"], 199, RANDOM_STATE).effect_size,
        "Diet R^2": permanova(D, metadata["Diet"], 199, RANDOM_STATE).effect_size,
    })

metric_table = pd.DataFrame(rows)
RESULTS["metric_comparison"] = metric_table.to_dict("records")
show_table(metric_table, "metric_comparison")
````

````python
fig, axes = plt.subplots(2, 3, figsize=(15, 8.6))
for ax, (name, _, _) in zip(axes.ravel(), metric_specs):
    ordination_plot(results_by_metric[name], labels=metadata["Donor"], ax=ax,
                    title=name, legend_title="Donor", point_size=14)
    ax.get_legend().remove()
axes.ravel()[0].legend(title="Donor", fontsize=7, ncol=2)
plt.tight_layout(); save(fig, "fig07_metric_comparison"); plt.show()
````

````python
names = list(distances)
M = np.eye(len(names))
for i in range(len(names)):
    for j in range(i + 1, len(names)):
        r = np.corrcoef(distances[names[i]][np.triu_indices(675, 1)],
                        distances[names[j]][np.triu_indices(675, 1)])[0, 1]
        M[i, j] = M[j, i] = r

fig, ax = plt.subplots(figsize=(5.6, 4.8))
im = ax.imshow(M, cmap="RdYlBu_r", vmin=0, vmax=1)
ax.set_xticks(range(len(names)), names, rotation=45, ha="right")
ax.set_yticks(range(len(names)), names)
for i in range(len(names)):
    for j in range(len(names)):
        ax.text(j, i, f"{M[i,j]:.2f}", ha="center", va="center", fontsize=8,
                color="white" if M[i, j] > 0.6 else "black")
ax.set_title("Mantel correlation between dissimilarity measures"); ax.grid(False)
plt.colorbar(im, shrink=0.8)
plt.tight_layout(); save(fig, "fig08_mantel"); plt.show()
````

Three findings worth carrying into the report.

1. **The metrics disagree, and the disagreement is systematic.** Bray–Curtis, Hellinger and
   Jensen–Shannon agree strongly with each other — all three read abundance profiles on a
   comparable scale. Jaccard and Aitchison sit apart: Jaccard throws abundance away entirely,
   and Aitchison works in log-ratio space where a taxon present at $10^{-5}$ counts as much
   as one at $10^{-1}$.
2. **Abundance carries the signal, not membership.** Every abundance-weighted metric
   recovers a much larger `Donor` effect than Jaccard does. Donors do not host disjoint sets
   of taxa; they host the *same* taxa in reliably different proportions.
3. **Euclidean distance appears to do best on % explained — and that is exactly the trap.**
   39% on two axes only looks good because raw Euclidean distance is dominated by the handful
   of most abundant taxa, so a low-dimensional fit is easy. It recovers *less* group structure
   (`Donor` $R^2$ 0.21 vs 0.29) while reporting a bigger number. Percentage explained measures
   how compressible the distance matrix is, not how biologically meaningful it is.

## 6 · Negative eigenvalues and additive-constant corrections

Bray–Curtis put 11% of the absolute trace into negative eigenvalues. Two classical fixes add
a constant to the dissimilarities so the corrected matrix becomes Euclidean-embeddable:

* **Lingoes (1971)** — $D' = \sqrt{D^2 + 2c}$, with $c = |\lambda_{\min}|$ of $\mathbf{G}$.
* **Cailliez (1983)** — $D' = D + c$ off-diagonal, where $c$ is the largest real eigenvalue
  of a $2N \times 2N$ block matrix built from $\mathbf{G}(D)$ and $\mathbf{G}(D^2)$.

Both are implemented from scratch. Cailliez is markedly more expensive: it needs a
*non-symmetric* eigendecomposition of a matrix twice the size, i.e. $O((2N)^3)$ with the
general solver rather than the symmetric one.

````python
rows = []
for corr in [None, "lingoes", "cailliez"]:
    t0 = time.perf_counter()
    r = pcoa.pcoa(D_bc, correction=corr)
    elapsed = time.perf_counter() - t0
    rows.append({
        "correction": corr or "none",
        "constant c": r.correction_constant,
        "time (s)": elapsed,
        "negative %": 100 * r.negative_fraction,
        "PCo1+2 %": 100 * r.proportion_explained[:2].sum(),
        "axes for 80%": int(np.searchsorted(np.cumsum(r.proportion_explained), 0.80) + 1),
        "Donor R^2 (2 axes)": permanova(
            pairwise_distances(r.coordinates[:, :2], "euclidean"),
            metadata["Donor"], 199, RANDOM_STATE).effect_size,
    })
correction_table = pd.DataFrame(rows)
RESULTS["corrections"] = correction_table.to_dict("records")
show_table(correction_table, "corrections")
````

**The corrections work, and we do not use them.** Both drive the negative fraction to
zero, but they do it by adding a constant to *every* dissimilarity, which inflates the total
variation with a term that carries no information. The share explained by the first two axes
collapses from 28% to 7% (Lingoes) or 15% (Cailliez), and the number of axes needed to reach
80% rises sharply — the structure is not lost, it is diluted across many low-information
dimensions.

The practical rule: correct when you need the embedding to be *metrically* valid
(for downstream methods that assume Euclidean geometry, or when negative eigenvalues rival
the leading positive ones). For visualisation and for PERMANOVA — which works from
$\mathbf{G}$ directly and is well defined for indefinite $\mathbf{G}$ — leave them in and
report the magnitude honestly. Here the largest negative eigenvalue is a small fraction of
$\lambda_1$, so the leading axes are trustworthy.

````python
neg = res_bc.eigenvalues[res_bc.eigenvalues < 0]
print(f"largest positive eigenvalue : {res_bc.eigenvalues[0]:.3f}")
print(f"most negative eigenvalue    : {neg.min():.3f} "
      f"({abs(neg.min())/res_bc.eigenvalues[0]:.1%} of lambda_1)")

fig, ax = plt.subplots(figsize=(7.2, 3.4))
ev = res_bc.eigenvalues
ax.bar(np.arange(len(ev)), ev, width=1.0,
       color=np.where(ev >= 0, "#4C72B0", "#C44E52"))
ax.axhline(0, color="black", lw=0.8)
ax.set(xlabel="eigenvalue index", ylabel="eigenvalue",
       title="Full eigenvalue spectrum: positive (blue) vs negative (red)")
plt.tight_layout(); save(fig, "fig09_spectrum"); plt.show()
````

## 7 · Evaluation

### 7.1 How faithful is the embedding?

Kruskal's stress-1 measures the gap between the original dissimilarities and the distances in
the reduced space:
$S_1 = \sqrt{\sum_{i<j}(d_{ij} - \hat d_{ij})^2 \,/\, \sum_{i<j} d_{ij}^2}$.

````python
ks = [k for k in [2, 3, 5, 10, 20, 50, 100, 200, 300] if k <= res_full.n_axes]
rows = []
for k in ks:
    Dk = res_full.reconstructed_distances(k)
    err = reconstruction_error(D_bc, Dk)
    shep = shepard_correlation(D_bc, Dk)
    rows.append({"axes": k, "stress-1": kruskal_stress(D_bc, Dk),
                 "relative Frobenius": err["relative_frobenius"], "RMSE": err["rmse"],
                 "Shepard r": shep["pearson_r"], "Spearman rho": shep["spearman_rho"],
                 "cumulative %": 100 * res_full.proportion_explained[:k].sum()})
fidelity = pd.DataFrame(rows)
RESULTS["fidelity"] = fidelity.to_dict("records")
show_table(fidelity, "embedding_fidelity")
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
shepard_diagram(D_bc, res_full.reconstructed_distances(2), ax=axes[0],
                title="Shepard diagram, 2 axes")
axes[1].plot(fidelity["axes"], fidelity["stress-1"], "o-", color="#4C72B0", label="stress-1")
axes[1].plot(fidelity["axes"], fidelity["Shepard r"], "s-", color="#55A868", label="Shepard r")
for y, lbl in [(0.20, "fair"), (0.10, "good"), (0.05, "excellent")]:
    axes[1].axhline(y, color="grey", ls=":", lw=1)
    axes[1].text(fidelity["axes"].max(), y + 0.006, lbl, fontsize=7.5, color="grey", ha="right")
axes[1].set(xscale="log", xlabel="number of principal coordinates", ylabel="value",
            title="Fidelity vs dimensionality")
axes[1].legend()
plt.tight_layout(); save(fig, "fig10_shepard_stress"); plt.show()
````

````python
best_k = int(fidelity.loc[fidelity["stress-1"].idxmin(), "axes"])
print(f"stress-1 is minimised at {best_k} axes "
      f"(S1 = {fidelity['stress-1'].min():.3f}); it rises again beyond that")
RESULTS["best_k_by_stress"] = best_k
RESULTS["stress_2d"] = float(fidelity.loc[fidelity["axes"] == 2, "stress-1"].iloc[0])
````

Stress at two axes is high (0.514), and the Shepard diagram shows why: the cloud sits
*below* the 1:1 line almost everywhere. A 2-D map systematically **under-states** true
dissimilarities, because structure that needs hundreds of dimensions is compressed into two.
The conventional "good" threshold of 0.10 is first reached at **20 axes** — which, not by
coincidence, is exactly what the broken-stick rule retained in section 3.

**Stress does not decrease monotonically.** It bottoms out at 0.036 around 50 axes and then
*rises* again — 0.052 at 100 axes, 0.080 at 300. For a Euclidean input that would be
impossible. It happens here because Bray–Curtis is not Euclidean-embeddable: the negative
eigenvalues represent structure that no real configuration can reproduce, so beyond a certain
point each additional positive axis adds length that the original dissimilarities do not have,
and the reconstruction starts to *over*-state distances. The non-Euclideanity measured in
section 3 therefore has a direct practical consequence: there is an optimal number of axes,
and it is not "all of them".

This is the honest limitation of any 2-D microbiome ordination and it should temper how the
plot is read: *relative* positions and group centroids are meaningful, but the absolute gap
between two points on the page is not a reliable estimate of how different those two
communities really are. Note that the Spearman correlation stays high even where stress is
high — the *ordering* of dissimilarities survives projection far better than their magnitudes,
which is exactly the property that makes the ordination useful despite the compression.

### 7.2 Is the visible grouping statistically real?

Two complementary permutation tests, 999 permutations, seeded:

* **PERMANOVA** (Anderson 2001) partitions the sum of squares of $\mathbf{D}$ and compares
  between- to within-group variation with a pseudo-$F$. $R^2$ is the fraction of multivariate
  variation the factor explains. Sensitive to differences in group *dispersion*.
* **ANOSIM** (Clarke 1993) works on dissimilarity *ranks*, so it is insensitive to the
  distance scale and provides a check on PERMANOVA.

````python
rows = []
for f in FACTORS:
    g = metadata[f].to_numpy()
    pm = permanova(D_bc, g, N_PERMUTATIONS, RANDOM_STATE)
    an = anosim(D_bc, g, N_PERMUTATIONS, RANDOM_STATE)
    sil = silhouette_from_distances(D_bc, g)
    ari = cluster_agreement(res_bc.coordinates[:, :5], g, random_state=RANDOM_STATE)
    rows.append({"factor": f, "levels": metadata[f].nunique(),
                 "pseudo-F": pm.statistic, "R^2": pm.effect_size, "PERMANOVA p": pm.p_value,
                 "ANOSIM R": an.statistic, "ANOSIM p": an.p_value,
                 "silhouette": sil["mean"], "k-means ARI": ari["adjusted_rand_index"]})
stats_table = pd.DataFrame(rows).sort_values("R^2", ascending=False)
RESULTS["group_tests"] = stats_table.to_dict("records")
show_table(stats_table, "group_tests")
````

Read this table row by row.

* **`Donor` explains 28.7% of the multivariate variation** — twice as much as anything else,
  with an ANOSIM $R$ of 0.49, the only factor approaching genuine separation. Host identity
  dominates.
* **`Diet` is a real but secondary effect**: $R^2 = 0.14$, $p = 0.001$, ANOSIM $R = 0.31$.
* **`Sex` is the cautionary row.** PERMANOVA calls it significant ($p = 0.001$) while ANOSIM
  returns $R = -0.06$ with $p \approx 0.95$ — no separation whatsoever. With 621 vs 54
  samples, a difference in within-group *dispersion* alone is enough to produce a significant
  pseudo-$F$. Reporting the PERMANOVA result without the rank-based check would have been a
  false positive. This is why both tests are run.
* **Every silhouette is near zero** despite highly significant $p$-values. There is no
  contradiction: significance says the group *centroids* differ; silhouette says individual
  samples still sit about as close to other groups as to their own. Microbiome groups
  overlap heavily. Any claim of "clusters" here would be an over-reading — these are
  *gradients* with displaced centres.

The $p$-value floor of $1/(999+1) = 0.001$ is a property of the permutation scheme, not a
measure of how extreme the result is; the effect sizes carry that information.

````python
# sensitivity check: Diet levels 2 and 3 hold one sample each
main = metadata["Diet"].isin([0, 1]).to_numpy()
D_sub = D_bc[np.ix_(main, main)]
pm = permanova(D_sub, metadata.loc[main, "Diet"], N_PERMUTATIONS, RANDOM_STATE)
an = anosim(D_sub, metadata.loc[main, "Diet"], N_PERMUTATIONS, RANDOM_STATE)
print(f"Diet restricted to the two populated levels (n={main.sum()}):")
print(f"  {pm}\n  {an}")
RESULTS["diet_sensitivity"] = {"n": int(main.sum()), "R2": pm.effect_size,
                               "p": pm.p_value, "anosim_R": an.statistic}
````

The `Diet` effect survives dropping the singleton levels, so it is not an artefact of
degenerate groups.

## 8 · Performance

Three questions: where does the time go, how does it scale, and can we do better than a full
eigendecomposition when only a few axes are needed?

````python
def _time(fn):
    """Wall-clock a callable once."""
    t0 = time.perf_counter(); fn(); return time.perf_counter() - t0

sizes = [50, 100, 200, 350, 500, 675]
rng = np.random.default_rng(RANDOM_STATE)
rows = []
for n in sizes:
    idx = rng.choice(675, n, replace=False)
    Xs = X_count[idx]
    repeats = 5 if n <= 200 else 2          # small n is dominated by fixed overhead
    t_dist = min(_time(lambda: pairwise_distances(Xs, "braycurtis")) for _ in range(repeats))
    Ds = pairwise_distances(Xs, "braycurtis")
    t_pcoa = min(_time(lambda: pcoa.pcoa(Ds, n_components=10)) for _ in range(repeats))
    tracemalloc.start(); pcoa.pcoa(Ds, n_components=10)
    peak = tracemalloc.get_traced_memory()[1] / 1e6
    tracemalloc.stop()
    rows.append({"n samples": n, "distance (s)": t_dist, "PCoA (s)": t_pcoa,
                 "total (s)": t_dist + t_pcoa, "peak memory (MB)": peak,
                 "D matrix (MB)": Ds.nbytes / 1e6})
scaling = pd.DataFrame(rows)
RESULTS["scaling"] = scaling.to_dict("records")
show_table(scaling, "scaling")
````

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.0))
n = scaling["n samples"].to_numpy()

axes[0].loglog(n, scaling["distance (s)"], "o-", color="#4C72B0", label="distance  $O(N^2P)$")
axes[0].loglog(n, scaling["PCoA (s)"], "s-", color="#C44E52", label="eigendecomp.  $O(N^3)$")
ref2 = scaling["distance (s)"].iloc[-1] * (n / n[-1]) ** 2
ref3 = scaling["PCoA (s)"].iloc[-1] * (n / n[-1]) ** 3
axes[0].loglog(n, ref2, ":", color="#4C72B0", alpha=0.6)
axes[0].loglog(n, ref3, ":", color="#C44E52", alpha=0.6)
axes[0].set(xlabel="samples N", ylabel="seconds", title="Runtime scaling (dotted = theory)")
axes[0].legend()

axes[1].plot(n, scaling["peak memory (MB)"], "o-", color="#55A868", label="measured peak")
axes[1].plot(n, scaling["D matrix (MB)"], "s--", color="#8172B3", label="$D$ matrix alone")
axes[1].set(xlabel="samples N", ylabel="MB", title="Memory footprint  $O(N^2)$")
axes[1].legend()

# solver comparison at full size
solver_rows = []
for solver, k in [("full", None), ("partial", 10), ("partial", 3)]:
    t0 = time.perf_counter()
    r = pcoa.pcoa(D_bc, n_components=k, eigen_solver=solver)
    solver_rows.append({"solver": f"{solver} (k={k or 'all'})", "time (s)": time.perf_counter() - t0,
                        "lambda_1": r.eigenvalues[0]})
solver_df = pd.DataFrame(solver_rows)
axes[2].barh(solver_df["solver"], solver_df["time (s)"], color=["#4C72B0", "#DD8452", "#55A868"])
axes[2].set(xlabel="seconds", title="Eigensolver: full vs partial")
axes[2].grid(axis="y", alpha=0)
for i, v in enumerate(solver_df["time (s)"]):
    axes[2].text(v, i, f" {v:.3f}s", va="center", fontsize=9)
plt.tight_layout(); save(fig, "fig11_performance"); plt.show()

RESULTS["solvers"] = solver_df.to_dict("records")
assert np.allclose(solver_df["lambda_1"].to_numpy(), solver_df["lambda_1"].iloc[0], rtol=1e-6)
print("all solvers agree on lambda_1 to 1e-6")
````

**Where the time goes.** At $N = 675$ the distance computation ($O(N^2P)$, ~9 s)
dominates the eigendecomposition ($O(N^3)$, ~0.1 s) by two orders of magnitude — the
opposite of the usual assumption. The measured slopes sit slightly *below* the theoretical
$N^2$ and $N^3$ references, which is expected: at these sizes both routines are still partly
overhead- and cache-bound rather than FLOP-bound, and LAPACK's blocked eigensolver amortises
its fixed costs as $N$ grows. The asymptotic crossover is at roughly $N \approx P$: beyond a
few thousand samples the eigendecomposition takes over and PCoA itself becomes the bottleneck.

**Memory is the real wall.** The $O(N^2)$ distance matrix is 3.6 MB here, but at
$N = 100{,}000$ samples it would be 80 GB — which is why our distance routines are blocked
against an explicit memory budget rather than materialising the $N \times N \times P$
broadcast array (24 GB even for this modest dataset).

**Partial eigensolvers are the scalability lever.** Lanczos iteration (`eigsh`) extracts only
the leading $k$ eigenpairs at $O(N^2k)$ instead of $O(N^3)$, and already wins at this size: 3
axes in 0.015 s against 0.066 s for the full decomposition, agreeing on $\lambda_1$ to
$10^{-6}$. The gap widens cubically with $N$. The trade-off is that a partial solve returns
only the leading end of the spectrum, so the negative eigenvalues — and therefore the
non-Euclideanity diagnostic of section 6 — are invisible to it. That is why the full solver
remains the default here and the partial one is the escape hatch for large $N$.

## 9 · Extension — covariate-adjusted PCoA

Section 4 left us with a problem the standard algorithm cannot solve: `Donor` dominates the
ordination, and because donors are unevenly distributed across diets, it also *confounds* the
`Diet` comparison. This is precisely the motivating case of Shi *et al.* (2020), the paper
this dataset comes from.

Their fix, aPCoA, projects the covariate out of Gower's matrix before the decomposition. With
$\mathbf{H} = \mathbf{X}(\mathbf{X}'\mathbf{X})^{-1}\mathbf{X}'$ the hat matrix of the
nuisance covariates,

$$\boldsymbol{\Delta}^{*} = (\mathbf{I} - \mathbf{H})\,\mathbf{G}\,(\mathbf{I} - \mathbf{H})$$

and $\boldsymbol{\Delta}^{*}$ is then decomposed exactly as before. Implementing this on top
of our PCoA is a five-line change, which is the best argument for having written the algorithm
modularly in the first place.

````python
covariates = dummy_code(metadata, ["Donor"])
res_adj = apcoa(D_bc, covariates, n_components=10)

def factor_r2_in_plane(result, factor, k=2):
    """PERMANOVA R^2 computed on the first k principal coordinates only."""
    D_plane = pairwise_distances(result.coordinates[:, :k], metric="euclidean")
    return permanova(D_plane, metadata[factor], 199, RANDOM_STATE).effect_size

rows = []
for f in ["Donor", "Diet", "Source", "Sex"]:
    rows.append({"factor": f,
                 "R^2 in PCo1-2 (standard)": factor_r2_in_plane(res_bc, f),
                 "R^2 in PCo1-2 (adjusted for Donor)": factor_r2_in_plane(res_adj, f)})
adj_table = pd.DataFrame(rows)
RESULTS["apcoa"] = adj_table.to_dict("records")
show_table(adj_table, "apcoa_comparison")
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12.6, 5.0))
ordination_plot(res_bc, labels=metadata["Diet"], ax=axes[0], legend_title="Diet",
                title="Standard PCoA — Diet effect masked by Donor", point_size=16)
ordination_plot(res_adj, labels=metadata["Diet"], ax=axes[1], legend_title="Diet",
                title="aPCoA adjusted for Donor — Diet effect exposed", point_size=16)
plt.tight_layout(); save(fig, "fig12_apcoa"); plt.show()
````

After adjustment the `Donor` signal in the leading plane is gone by construction
($R^2 \to 0$), while `Diet` retains — and relative to the remaining variation, strengthens —
its structure. The plot on the right is the one to show when the scientific question is about
diet, and the plot on the left is the one to show when asking why the experiment is hard.
Neither is "the" ordination; each answers a different question, and saying which question is
being asked is part of reporting the result.

One caveat: aPCoA removes only the *linear* projection of the covariate. It is a visualisation
aid, and the inferential claim about diet still belongs to a PERMANOVA that includes donor as a
term — not to the adjusted picture.

## 10 · Validation

Three independent lines of evidence that the implementation is correct.

**(a) The algebraic identity.** PCoA on Euclidean distances must equal PCA on the centred
data, with eigenvalues $(n-1)\times$ the PCA explained variances. This is a strong test: the
two routines share no code.

````python
from sklearn.decomposition import PCA

D_euc = distances["euclidean"]
res_euc = pcoa.pcoa(D_euc, n_components=5)
pca = PCA(n_components=5).fit(X_count)
scores = pca.transform(X_count)

print(f"Procrustes m^2 (PCoA vs PCA scores)  : {procrustes_m2(res_euc.coordinates, scores):.3e}")
print(f"max relative eigenvalue difference   : "
      f"{np.abs(res_euc.eigenvalues[:5]/(pca.explained_variance_*(len(X_count)-1)) - 1).max():.3e}")
````

**(b) Agreement with scikit-bio**, an independent reference PCoA implementation.

````python
try:
    import skbio
    ref = skbio.stats.ordination.pcoa(skbio.DistanceMatrix(D_bc), number_of_dimensions=5)
    ours = pcoa.pcoa(D_bc, n_components=5)
    print(f"scikit-bio {skbio.__version__}")
    print(f"max relative eigenvalue difference : "
          f"{np.abs(ours.eigenvalues[:5]/ref.eigvals.values[:5] - 1).max():.3e}")
    print(f"Procrustes m^2 between coordinates : "
          f"{procrustes_m2(ours.coordinates, ref.samples.values[:, :5]):.3e}")
    RESULTS["skbio_validated"] = True
except ImportError:
    print("scikit-bio not installed - skipping (pip install scikit-bio)")
    RESULTS["skbio_validated"] = False
````

**(c) The test suite** — 36 cases covering metric identities, Gower centring, eigen-structure invariants, corrections, aPCoA, the evaluation statistics and error handling.

````python
import subprocess
out = subprocess.run([sys.executable, "-m", "pytest", "tests/", "-q", "--no-header"],
                     capture_output=True, text=True)
print(out.stdout.strip().splitlines()[-1])
RESULTS["tests_passed"] = "failed" not in out.stdout
````

## 11 · Conclusions

**On the algorithm.** PCoA is four lines of linear algebra whose correctness rests entirely on
one identity — that doubly centring $-\tfrac12 D^2$ recovers a Gram matrix. Verified on a
unit square, against the PCA identity, and against scikit-bio, our implementation reproduces
the reference answers to machine precision. The subtleties are not in the decomposition but
around it: what dissimilarity to feed it, what negative eigenvalues mean, and how much of the
picture survives projection to two dimensions.

**On the data.** Bray–Curtis PCoA of 675 samples × 2385 taxa puts 28.0% of the positive
variation on the first two axes, with 11.0% of the absolute trace in negative eigenvalues.
`Donor` is the dominant source of structure ($R^2 = 0.287$, ANOSIM $R = 0.49$), `Diet` a real
secondary effect ($R^2 = 0.140$) that survives dropping the singleton levels, and
`CollectionMet` and `Sex` are negligible. The `Sex` row is the methodological lesson of the
analysis: PERMANOVA called it significant and rank-based ANOSIM refuted it, a dispersion
artefact of a 621-vs-54 split that a single test would have reported as a finding.

**On limits.** Stress at two axes is 0.51, so the map compresses distances substantially and
absolute inter-point gaps on the page should not be over-read; 20 axes are needed to reach the
conventional 0.10 threshold, matching the broken-stick rule exactly. Stress then *rises* again
beyond ~50 axes — a direct, measurable consequence of the negative eigenvalues, and a reminder
that with a non-Euclidean dissimilarity "more dimensions" is not automatically "more faithful".
Near-zero silhouettes across every factor mean these are overlapping gradients, not clusters:
statistically distinct centroids, biologically continuous communities.

**On the extension.** Because donor identity both dominates the ordination and confounds the
diet comparison, we implemented the covariate-adjusted variant from the source paper. It
removes the donor gradient by construction and exposes the diet structure underneath — a
better picture for one specific question, and not a replacement for the unadjusted view.

**On performance.** At this scale the $O(N^2P)$ distance computation, not the $O(N^3)$
eigendecomposition, is the bottleneck, and $O(N^2)$ memory is the first thing that breaks as
$N$ grows. Blocked distance computation and a partial (Lanczos) eigensolver are the two levers
that extend the same code to substantially larger cohorts.
